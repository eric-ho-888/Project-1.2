```python
# 1. Import libraries + Load data

import pandas as pd
import numpy as np

df = pd.read_csv('data.csv')  # or pd.read_excel, pd.read_json, pd.read_sql

# Check what you loaded first

df.head()           # show first 5 rows
df.tail()           # show last 5 rows
df.info()           # show all information of each column
df.shape            # what is total no. of table (row, column)
df.dtypes           # what type of data
df.describe()       # stats for numeric column
```

```python
# 3. Explore + Feature Engineering (create new columns to answer questions)

df['revenue'] = df['qty'] * df['price']          # pandas
df['month'] = df['date'].dt.month

df['is_high_value'] = np.where(df['revenue'] > 1000, 1, 0) # numpy conditional
df['age_group'] = pd.cut(df['age'], bins=[0,18,35,60,100]) # binning
```

```python
# 4. Analyze + Aggregate (this is where pandas shines)

df.groupby('city')['revenue'].sum()          # sum by group
df.pivot_table(index='month', columns='city', values='revenue', aggfunc='mean')
df.corr()                                    # correlation matrix with numpy under the hood

# Numpy for calculations

np.percentile(df['price'], [25, 50, 75])     # quartiles
np.cov(df[['price', 'qty']].T)               # covariance
```

```python
# 6. Visualize (Pandas uses matplotlib under the hood)

df['revenue'].hist()                         # histogram
df.groupby('city')['revenue'].sum().plot(kind='bar') # bar chart
df.plot.scatter(x='price', y='qty')          # scatter
```
